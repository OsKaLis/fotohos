# <span style='color:RoyalBlue'>Описание проекта</span>

<span style='font-size:18px'>

Пользователи хостинга «<span style='color:green'>Со Смыслом</span>» размещают свои фотографии на хостинге и сопровождают их полным описанием:

- модель камеры и т. д.

Отличительная особенность сервиса — описание: его может предоставить не только тот, кто размещает фотографию, но и другие пользователи портала.

Отдел занимается экспериментом по разработке поиска референсных фотографий для фотографов. 
Суть поиска заключается в следующем: пользователь сервиса вводит описание нужной сцены.
Чтобы эксперимент получил право на жизнь, нужно защитить его перед руководителем компании. Для защиты необходимо презентовать так называемый PoC (Proof of Concept, Проверка концепции) — продемонстрировать, что такой проект практически осуществим.

### Задача: Создание финкции вывода первичной информации.

````python
def primary_information_output(dataset):
    # Функция вывод первичной информации
    # Первичная информация
    dataset.info()
    # О пропусках и их доля
    print('Количество пропущеных значений и их доля:')
    number_of_entries = dataset.shape[0]
    preliminary_frame = pd.DataFrame(
        {
            'Наименование признака': [],
            'Количество пропущеных': [],
            'Доля пропущеных': [],
        }
    )
    for key, value in dataset.isna().sum().items():
        preliminary_frame.loc[len(preliminary_frame)] = [
            key,
            value,
            (value / number_of_entries),
        ]
    display(preliminary_frame)
    # Информация о полных дублях
    number_of_complete_takes = dataset.duplicated().sum()
    percentage_of_complete_takes = number_of_complete_takes / number_of_entries
    text = 'Количество полных дублированых строк в дата фрейме'
    print(f'{text} = {number_of_complete_takes}, Доля полных дублей = {percentage_of_complete_takes:.1%}')
    # Вывод первых 5 строк фрейма
    display(dataset.head())
````

В этом проекте есть несколько возможных стратегий работы с обучающими и краудсорс оценками,  вот эти стратегии с плюсами и минусами:
   

   - использовать соединение через outer, в этом случае, при корректной обработке пропусков мы можем получить более 50К размеченых пар. К плюсам данного подхода можно отнести большое количество данных. Минус - в основном это будут данные, размеченные краудсорсерами, а там качество раметки ниже.
   - использовать только экспертные оценки. плюс - высокое качество данных (разметка имеет шкалу). Минус - данных меньше.  Поясню по поводу качества разметки на примере: если на изображении одна собака, а в описании "две собаки бегут по берегу", то эксперты поставять соответствие 0,6-0,7 (переводя в шкалу 0-1), а краудсорсеры поставят 0.  Но на экспертных оценках модель хотя-бы научится находить собак, а на краудсор оценках ничему не начится.

   - использование соединения через left.  Данный подход очевидно проигрывает второму варианту: данных столько же, нужно дополнительно возиться с объединением, но качество разметки снижается.. Но это соответсвует предложениям авторов проекта, поэтому такой подход имеет право на жизнь

   - объединение через inner. Тут все плохо - и данных ОЧЕНЬ мало (в два раза меньше, чем использовать только экспертные оценки) и качество у них "подпорчено". И навыков по обработке пропусков не получаем.....

### Задача: Векторизация текста и изображения.

````python
# Функции для обработки

#Mean Pooling - Take attention mask into account for correct averaging
def mean_pooling(model_output, attention_mask):
    token_embeddings = model_output[0] #First element of model_output contains all token embeddings
    input_mask_expanded = attention_mask.unsqueeze(-1).expand(token_embeddings.size()).float()
    return torch.sum(token_embeddings * input_mask_expanded, 1) / torch.clamp(input_mask_expanded.sum(1), min=1e-9)

def get_text_embedding(text):
    """
    Получение эмбеддинга текста через BERT (PyTorch)
    """
    # Tokenize sentences
    encoding = tokenizer(
        text,
        padding=True,
        truncation=True,
        return_tensors='pt'
    )
    # Compute token embeddings
    with torch.no_grad():
        output = bert_model(**encoding)
        # Perform pooling
        sentence_embeddings = mean_pooling(output, encoding['attention_mask'])
        # Normalize embeddings
        sentence_embeddings = F.normalize(sentence_embeddings, p=2, dim=1)
    return sentence_embeddings.squeeze().numpy() # конвертируем в numpy

def get_image_embedding(image_path):
    """
    Получение эмбеддинга изображения через ResNet18 (PyTorch)
    """
    # Загрузка предобученной модели ResNet18
    resnet18 = models.resnet18(pretrained=True)
    
    # Убираем последний полносвязный слой, чтобы получить эмбеддинг
    resnet18 = nn.Sequential(*list(resnet18.children())[:-1])
    resnet18.eval()  # Переводим модель в режим инференса

    # Трансформации для изображения (аналогично предобработке в ResNet)
    transform = transforms.Compose([
        transforms.Resize(256),
        transforms.CenterCrop(224),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406],
                           std=[0.229, 0.224, 0.225])
    ])

    # Загрузка и предобработка изображения
    image = Image.open(image_path).convert('RGB')
    image_tensor = transform(image)
    image_tensor = image_tensor.unsqueeze(0)  # Добавляем размерность батча

    # Получение эмбеддинга
    with torch.no_grad():
        embedding = resnet18(image_tensor)
        embedding = embedding.squeeze().numpy()  # Конвертируем в numpy
        # L2-нормализация
        embedding_norm = embedding / (np.linalg.norm(embedding) + 1e-10)

    return embedding_norm
````

Однако стоит учитывать, что не все энкодерные архитектуры обучались на задаче NSP (например, DistilBERT).

На практике часто используют пулинг (например, средний) по всем выходным токенам (кроме паддингов), чтобы получить более общее представление текста. Многие исследователи отмечают, что при этом качество эмбеддингов оказывается выше. Если встретишь такой подход в чужом коде (или туториале), не удивляйся.

Вот пример реализации такого подхода:
    

```python
def preprocess_text_with_pooling(text):
    inputs = tokenizer(text, return_tensors='pt', truncation=True, padding='max_length', max_length=128).to(device)
    attention_mask = inputs['attention_mask']
    with torch.no_grad():
        outputs = distilbert_model(**inputs)

    last_hidden_state = outputs.last_hidden_state

    # Убираем влияние паддинг-токенов
    attention_mask = attention_mask.unsqueeze(-1)
    masked_hidden_state = last_hidden_state * attention_mask
    sum_hidden_states = masked_hidden_state.sum(dim=1)
    sum_mask = attention_mask.sum(dim=1)
    mean_hidden_state = sum_hidden_states / sum_mask

    return mean_hidden_state
```

Такой подход позволяет учесть всю доступную информацию в тексте, исключая влияние токенов-паддингов, что иногда даёт более качественные эмбеддинги.

````python
def data_preparation(df, path_image):
    """
    Векторизация уникальных картинок и текста в единый датафрейм
    и распределения по всему дата фрейму
    """
    # Получаем уникальных значений для текста и картинки
    unique_images = df['image'].unique()
    unique_texts = df['query_text'].unique()
    # Обрабатываем каждую уникальную картинку и текст
    image_cache = {img_path: get_image_embedding(path_image + img_path) for img_path in unique_images}
    text_cache = {text: get_text_embedding(text) for text in unique_texts}
    
    # Разкладываем эмбеддинги картинок на столбцы
    image_embeddings = df['image'].map(image_cache)
    image_df = pd.DataFrame(
        image_embeddings.tolist(),
        index=df.index,
        columns=[f'img_emb_{i}' for i in range(len(image_embeddings.iloc[0]))]
    )
    
    # Разкладываем эмбеддинги текстов на столбцы
    text_embeddings = df['query_text'].map(text_cache)
    text_df = pd.DataFrame(
        text_embeddings.tolist(),
        index=df.index,
        columns=[f'text_emb_{i}' for i in range(len(text_embeddings.iloc[0]))]
    )
    
    # Объединяем с исходным датафреймом
    df = pd.concat([df, text_df, image_df], axis=1)
    return df
````

**1. Проблема «N+1 query» и дубликатов**
В текущей реализации функции `data_preparation` векторизация вызывается внутри цикла `iterrows`. 
* **Нюанс:** Многие изображения и текстовые описания в датафрейме дублируются. 
* **Последствие:** Модели (BERT и ResNet) вынуждены заново обрабатывать один и тот же контент несколько раз. Это создает колоссальную избыточную нагрузку на CPU/GPU и дисковую подсистему.
* **Рекомендация:** Рекомендуется сначала получить уникальные значения текстов и путей к изображениям, векторизовать их один раз и сохранить в кэш (или словарь), а затем «раздать» готовые эмбеддинги по строкам датафрейма.

**2. GPU-голодание и отсутствие батчинга**
Обработка объектов по одному — это антипаттерн в Deep Learning. 
* **Механика:** Пересылка тензора в память видеокарты для одного маленького вектора занимает больше времени, чем само вычисление. Видеокарта большую часть времени простаивает в ожидании. 
* **Рекомендация:** Рекомендуется использовать **батчинг (batch processing)**. Современные модели оптимизированы для обработки пакетов данных (например, по 32 или 64 объекта за раз). Это позволяет максимально эффективно использовать параллелизм GPU и ускоряет процесс в десятки раз.


**3. «Конфликт интересов» (Zoo of libraries)**
Использование одновременно **PyTorch** (для BERT) и **TensorFlow/Keras** (для ResNet) — рискованное решение для одного проекта. 
* **Риски:** Оба фреймворка по умолчанию стремятся захватить всю доступную видеопамять (VRAM) под свои нужды. Это прямой путь к ошибкам `Out of Memory` или непредсказуемым задержкам из-за конфликтов драйверов.  Да и в продакшене поддерживать образ с двумя библиотеками будет заметно сложнее.

* **Рекомендация:** Рекомендуется привести стек к единому стандарту. Так как BERT уже реализован на PyTorch, логичнее перенести и ResNet на этот же фреймворк (используя `torchvision.models`). Это сделает среду предсказуемой и облегчит деплой.

**4. Pandas-way: уход от `apply` и `iterrows`**
С точки зрения инженерии данных, использование `.iterrows()` и `.apply(axis=1)` считается неэффективным.

Пропущен важный шаг — масштабирование признаков. Это не просто рекомендация, а почти обязательное действие, отсутствие которого напрямую повлиет на твои результат, например нейронная сеть может обучаться медленнее и нестабильнее, также возмжоны проблемы с обычными линейными моделями.

Однако просто взять обычный `StandardScaler` здесь было бы ошибкой. Этот инструмент отлично подходит для обычных табличных данных, но он может «испортить» эмбеддинги, так как работает *поколоночно* и разрушает внутреннюю геометрию векторов, в которой и заключен весь их смысл.

Самый правильный и современный подход для твоей задачи — это **раздельная L2-нормализация** векторов перед их объединением.

Вот как это выглядит на практике:
1.  **Сначала** ты берешь все векторы изображений (из `ResNet`) и нормализуешь их отдельно с помощью `Normalizer(norm='l2')`.
2.  **Потом** ты берешь все векторы текстов (из `BERT`) и тоже нормализуешь их отдельно, своим экземпляром `Normalizer`'а.
3.  И только **после этого** ты конкатенируешь уже нормализованные векторы в единую матрицу признаков.

Что это дает? Ты приводишь оба типа векторов к сопоставимому масштабу (все они становятся векторами единичной длины), но при этом полностью сохраняешь семантическую структуру внутри каждого из пространств (пространства картинок и пространства текстов). Это самый надежный способ подготовить такие сложные признаки для модели.

### Задача: Создание класса для полносвязная нейронная сеть для регрессии. 

````python
class FullyConnectedModel(nn.Module):
    """
    Полносвязная нейронная сеть для регрессии с BatchNormalization
    """
    def __init__(self, input_dim, hidden_dims=[512, 128], dropout_rate=0.3, activation='relu'):
        super().__init__()
        self.network = self._build_network(input_dim, hidden_dims, dropout_rate, activation)

    def _build_network(self, input_dim, hidden_dims, dropout_rate, activation):
        """Создаёт последовательность слоёв сети"""
        layers = []
        dims = [input_dim] + hidden_dims

        for i in range(len(dims) - 1):
            layers.append(nn.Linear(dims[i], dims[i + 1]))
            layers.append(nn.BatchNorm1d(dims[i + 1]))  # Нормализация после линейного слоя
            layers.append(self._get_activation(activation))
            layers.append(nn.Dropout(dropout_rate))

        # Выходной слой (без активации, дропаута и нормализации)
        layers.append(nn.Linear(hidden_dims[-1], 1))
        layers.append(nn.Sigmoid())  # Гарантирует выход в [0, 1]

        return nn.Sequential(*layers)

    @staticmethod
    def _get_activation(act_name):
        """Возвращает функцию активации по имени"""
        activations = {
            'relu': nn.ReLU(),
            'sigmoid': nn.Sigmoid(),
            'tanh': nn.Tanh(),
            'leaky_relu': nn.LeakyReLU()
        }
        return activations.get(act_name, nn.ReLU())

    def forward(self, X=None, **kwargs):
        """
        Универсальный forward-метод:
        - принимает тензор X напрямую;
        - или объединяет данные из kwargs в один тензор.
        """
        if X is not None:
            return self.network(X)
        elif kwargs:
            # Объединяем все входные тензоры по последнему измерению
            values = list(kwargs.values())
            combined = torch.cat(values, dim=1)
            return self.network(combined)
        else:
            raise ValueError("Forward method requires either X or keyword arguments.")
````

**В чем заключается подвох:**

1.  **Дефицит данных для обучения:**
    * `GridSearchCV(cv=3)` делит данные так, что в каждый запуск `fit` попадает только **66%** датасета.
    * Внутри твоего метода `fit` вызывается `model.fit(..., validation_split=TEST_SIZE)`. Если `TEST_SIZE` равен, например, 0.2, то Keras отрезает еще 20% от тех 66%, что он получил.
    * **Итог:** твоя нейронка фактически обучается всего на **~53%** от общего объема данных. Для Deep Learning, где данных часто и так не хватает, это критическая потеря. Модель может просто не «доучиться» до нужного обобщения.

2.  **Конфликт интересов EarlyStopping и GridSearch:**
    * Коллбэк `EarlyStopping` мониторит `val_loss` на внутреннем «огрызке» данных, который отрезал Keras.
    * А `GridSearchCV` оценивает качество модели по внешнему валидационному фолду (тем 33%, которые он припас изначально).
    * Эти два набора данных не синхронизированы. Модель может остановиться по EarlyStopping, потому что она «плохо себя чувствует» на внутреннем куске, хотя на внешнем фолде она могла бы показать результат лучше. Это делает результаты поиска гиперпараметров нестабильными и шумными (высокая дисперсия метрик).

**В чем суть проблемы:**
В нашем датасете одно и то же изображение может встречаться несколько раз с разными описаниями. 
* Когда ты запускаешь стандартный `GridSearchCV` с `cv=3`, библиотека перемешивает строки случайным образом. 
* В итоге получается ситуация, когда одна пара «картинка А — описание 1» попадает в обучающий фолд, а пара «картинка А — описание 2» — в валидационный.
* **Результат:** Модель не «учится» искать смысл, она просто «узнает» картинку, которую уже видела в обучении. Это приводит к фиктивно высоким метрикам на валидации, которые мгновенно обрушатся на реальных, абсолютно новых изображениях.


**Как это исправить (Pandas & Scikit-learn way):**
Рекомендую использовать специализированный инструмент — **`GroupKFold`**. 

1. Нужно передать столбец с идентификаторами изображений (`image` или `group_column`) в метод `fit`  `GridSearchCV`.
2. В самом `GridSearchCV` в параметре `cv` укажи объект `GroupKFold`.

Это гарантирует, что все строки, относящиеся к одному и тому же изображению, всегда будут находиться либо целиком в обучении, либо целиком в валидации. Только так ты сможешь честно проверить, научилась ли твоя нейронка сопоставлять смыслы, а не просто запоминать пиксели.

Юрий, логика «один запрос ко всем изображениям» реализована верно: ты берешь текстовый вектор и сопоставляешь его со всей базой картинок для ранжирования. Однако с инженерной точки зрения процесс подготовки данных для модели можно сделать значительно быстрее и элегантнее.

**1. Уход от циклов (Broadcasting)**
Внутри функции `image_search_by_query` ты используешь цикл `for` для заполнения матрицы тестовым эмбеддингом. В мире высокопроизводительных вычислений и NumPy это считается антипаттерном.
* **Почему это важно:** Циклы в Python работают медленно. NumPy же оптимизирован для векторных операций на уровне C. 
* **Решение:** Используй **Broadcasting**. Ты можешь присвоить значение вектора целому срезу матрицы одним движением.

**Пример оптимизации:**
```python
# Вместо цикла:
# for i in range(X_test.shape[0]):
#     X_test_copy[i, :text_emb.shape[0]] = text_emb

# Используем NumPy-way:
X_test_copy[:, :text_emb.shape[0]] = text_emb
```
Это сработает мгновенно, так как NumPy сам «растянет» твой текстовый вектор по всем строкам матрицы на уровне скомпилированного кода.



**2. Архитектурная гибкость и работа с памятью**
Сейчас твоя функция жестко завязана на структуру `X_test` и делает её полную копию через `.copy()`.
* **Память:** Если в базе будет не 100, а 100 000 изображений, копирование такой матрицы на каждый запрос приведет к неоправданному расходу RAM.
* **Масштабируемость:** Логичнее хранить эмбеддинги изображений отдельно (как «базу данных»). При поиске ты просто берешь вектор запроса и конкатенируешь его с матрицей картинок «на лету». Это позволит добавлять новые фото в систему без пересборки тестовых датасетов.

**3. Векторный результат**
Вместо того чтобы оборачивать предсказания в `pd.Series` и делать `pd.concat`, эффективнее работать с индексами в NumPy. Метод `y_pred.argsort()` вернет тебе индексы самых релевантных картинок гораздо быстрее, чем сортировка целого DataFrame.

**Итог:** Твоя поисковая логика абсолютно рабочая, но «переход на рельсы» векторизации сделает её по-настоящему быстрой и профессиональной. Попробуй избавиться от этого цикла — это хорошая практика для любого инженера, работающего с данными.
